# **Data Visualization**

```python
import matplotlib.pyplot as plt

mujtaba['Exited'].value_counts().sort_index().plot(kind='bar', color='b')

plt.xlabel('Exited (0 = No, 1 = Yes)')
plt.ylabel('Count')
plt.title('Customer Churn Record')
plt.show()
```

```python
plt.bar(nadeem.index, nadeem.values, color='r')
plt.xlabel('index numbers')
plt.ylabel('corresponding values')
plt.title('Cstomer Churn Record')
```

```python
mujtaba.groupby('Geography')['Exited'].value_counts().unstack().plot(kind='bar')
plt.title('Churn by Geography')
plt.ylabel('Count')
plt.show()
```

```python
mujtaba['Exited'].value_counts().plot(kind='pie', autopct='%1.1f%%', colors=['green','red'])
plt.title('Churn Percentage')
plt.ylabel('')
plt.show()
```
